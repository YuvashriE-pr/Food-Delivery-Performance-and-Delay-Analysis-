## BeautifulSoup Web Scraping

```python
# Importing required packages

import requests
from bs4 import BeautifulSoup
import pandas as pd
```

```python
url = "https://en.wikipedia.org/wiki/List_of_cities_in_India_by_population"

headers = {
    "User-Agent": "FoodDeliveryAnalysis/1.0 (student project)"
}

response = requests.get(url, headers=headers)

print(response.status_code)
```

```python
from bs4 import BeautifulSoup
soup = BeautifulSoup(response.text, "html.parser")
print(soup.title.text)
```

```python
tables = soup.find_all("table")

print("Number of tables:", len(tables))
```

```python
for i, table in enumerate(tables):
    print("TABLE:", i)
    print(table.get_text(" ", strip=True)[:300])
    print("-" * 80)
```

```python
city_table = tables[1]

rows = city_table.find_all("tr")

print("Number of rows:", len(rows))
```

```python
city_data = []

for row in rows[1:21]:
    columns = row.find_all("td")
    
    if len(columns) >= 4:
        city = columns[0].get_text(strip=True)
        state = columns[2].get_text(strip=True)
        population = columns[3].get_text(strip=True)
        
        city_data.append([city, state, population])
```

```python
external_city_df = pd.DataFrame(
    city_data,
    columns=["City", "State", "Population"]
)

print(external_city_df)
```

```python
external_city_df.to_csv(
    "external_city_data.csv",
    index=False
)

print("File exported successfully!")
```
