### Step 2: Data Cleaning

```python
#Checking missing values 
df.isna().sum()
```

```python
df[df["delivery_time"].isnull()].head()
```

```python
#  delivery_time - 1039 missing,The order was cancelled, so the food was never delivered so,there is no delivery time
# delivery_duration_min - 1039 missing ,Since the order was cancelled, there was no completed delivery.
# cancel_reason — 8961 missing 
# successful orders , is_canceled = False and cancel_reason = NaN
# The order was not cancelled, so there is no cancellation reason
# Replacing missing values of cancel_reason with 'Not Cancelled'
# customer_rating — 1039 missing(is_canceled = True,customer_rating = NaN)
# The order was cancelled, so there was likely no completed customer experience to rate

df["cancel_reason"] = df["cancel_reason"].fillna("Not Cancelled")
```

```python
# Checking if value is correctly replaced
df.isnull().sum()
```

```python
# Ensuring Null value is correctly replaced 
df.tail()
```

```python
# Checking duplicate values
df.duplicated().sum()
```

```python
# Assisgning correct datatype to order_time,delivery_time,delivery_duration_min 
df["order_time"] = pd.to_datetime(df["order_time"])
```

```python
df["delivery_time"] = pd.to_datetime(df["delivery_time"])
```

```python
df["delivery_duration_min"] = df["delivery_duration_min"].astype(float)
```

```python
# Ensuring that the datatype of those column mentioned above is converted
df.dtypes
```

### Step 3: Create a New Column

```python
import numpy as np
df["Delivery_Status"] = np.where(
    df["delivery_duration_min"] <= 30,
    "Fast",
    np.where(
        df["delivery_duration_min"] <= 45,
        "Normal",
        "Delayed"
    )
)
df.head()
```

### Step 4: Perform Basic Analysis

```python
# Total number of orders
total_orders = df["order_id"].count()
print("Total number of orders:", total_orders)
```

```python
# Average delivery time 
average_delivery_time = df["delivery_duration_min"].mean()
print("Average delivery time:", round(average_delivery_time, 2))
```

```python
# Cancellation percentage
cancellation_percentage = df["is_canceled"].mean() * 100
print("Cancellation Percentage:", round(cancellation_percentage, 2), "%")
```

```python
# Average customer rating
average_rating = df["customer_rating"].mean()
print("Average Customer Rating:", round(average_rating, 2))
```

```python
#  Restaurant with the highest average delivery time
restaurant_avg_delivery = df.groupby("restaurant")["delivery_duration_min"].mean()

restaurant_highest_delivery = restaurant_avg_delivery.idxmax()

print("Restaurant:", restaurant_highest_delivery)
print("Average Delivery Time:", restaurant_avg_delivery.max())
```

```python
# Location with the highest number of delayed orders
delayed_orders = df[df["Delivery_Status"] == "Delayed"]
delayed_by_location = delayed_orders.groupby("city").size()
print("Location with highest delayed orders:", delayed_by_location.idxmax())
print("Number of delayed orders:", delayed_by_location.max())
```
